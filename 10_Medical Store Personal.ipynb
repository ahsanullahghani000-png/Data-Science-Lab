```python
# Count missing values in each column
print(df.isnull().sum())
```

```python
# Make a copy for basic imputation
df_basic = df.copy()

# Fill missing age and bmi with the median
df_basic["age"] = df_basic["age"].fillna(df_basic["age"].median())
df_basic["bmi"] = df_basic["bmi"].fillna(df_basic["bmi"].median())

# Fill missing region with the most common value (mode)
df_basic["region"] = df_basic["region"].fillna(df_basic["region"].mode()[0])

# Check that no missing value is left
print(df_basic.isnull().sum())
```

```python
# Numerical columns for advanced imputation
num_cols = ["age", "bmi", "children", "charges"]

# Make a copy of the data
df_knn = df.copy()

# KNN fills a missing value using the 5 most similar rows
knn = KNNImputer(n_neighbors=5)
df_knn[num_cols] = knn.fit_transform(df_knn[num_cols])

print("Missing values left:", df_knn[num_cols].isnull().sum().sum())
df_knn.head()
```

```python
# Make another copy of the data
df_iter = df.copy()

# Iterative Imputer predicts a missing value using the other columns
iterative = IterativeImputer(max_iter=10, random_state=42)
df_iter[num_cols] = iterative.fit_transform(df_iter[num_cols])

print("Missing values left:", df_iter[num_cols].isnull().sum().sum())
df_iter.head()
```

***Find the Residuals***

```python
# Residual = real value - predicted value (the error of the model)
residuals = y_train - simple_train_pred

print("Mean of residuals:", round(residuals.mean(), 2))

```

***Assumption 1: Linearity***

```python
# The points should follow a straight line
plt.figure(figsize=(8, 5))
sns.regplot(x=X_train["age"], y=y_train, line_kws={"color": "red"})
plt.title("Linearity Check: Age vs Charges")
plt.show()
```

***Assumption 2: Independence of Errors (Durbin-Watson)***

```python
# Value near 2 = errors are independent
# Near 0 or near 4 = errors are related to each other
dw = durbin_watson(residuals)
print("Durbin-Watson:", round(dw, 3))
```

***Assumption 3: Equal Variance (Homoscedasticity)***

```python
# Residuals vs predicted values
# The points should be spread evenly around the red line (no funnel shape)
plt.figure(figsize=(8, 5))
plt.scatter(simple_train_pred, residuals, alpha=0.6)
plt.axhline(0, color="red")
plt.title("Residuals vs Predicted Values")
plt.xlabel("Predicted Values")
plt.ylabel("Residuals")
plt.show()

# Breusch-Pagan test: p-value below 0.05 means variance is NOT equal
X_const = sm.add_constant(X_train_simple)
bp_test = het_breuschpagan(residuals, X_const)
print("Breusch-Pagan p-value:", bp_test[1])
```

***Assumption 4: Normality of Residuals***

```python
# Histogram and Q-Q plot of residuals
# For normal residuals: histogram looks like a bell, Q-Q points follow the red line
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].hist(residuals, bins=30)
axes[0].set_title("Histogram of Residuals")

stats.probplot(residuals, plot=axes[1])
axes[1].set_title("Q-Q Plot")

plt.tight_layout()
plt.show()

# Shapiro test: p-value below 0.05 means residuals are NOT normal
shapiro_test = stats.shapiro(residuals)
print("Shapiro p-value:", shapiro_test[1])
```

***Create Degree 4 Polynomial Features***

```python
# Degree 4 makes many new columns, so the model can easily overfit
poly4 = PolynomialFeatures(degree=4, include_bias=False)

X_train_poly4 = poly4.fit_transform(X_train_scaled)
X_test_poly4 = poly4.transform(X_test_scaled)

print("Columns before:", X_train_scaled.shape[1])
print("Columns after :", X_train_poly4.shape[1])
```

***Without Regularization (for comparison)***

```python
# Normal linear regression on degree 4 features (no penalty)
poly4_model = LinearRegression()
poly4_model.fit(X_train_poly4, y_train)

# Predict on train and test data
poly4_train_pred = poly4_model.predict(X_train_poly4)
poly4_pred = poly4_model.predict(X_test_poly4)

print("Train R2 :", round(r2_score(y_train, poly4_train_pred), 3))
print("Test R2  :", round(r2_score(y_test, poly4_pred), 3))
print("Test MAE :", round(mean_absolute_error(y_test, poly4_pred), 2))
print("Test RMSE:", round(np.sqrt(mean_squared_error(y_test, poly4_pred)), 2))
```

***Try Different Alpha Values (Ridge)***

```python
# Bigger alpha = stronger penalty
for alpha in [0.01, 0.1, 1, 10, 100, 1000]:
    model = Ridge(alpha=alpha)
    model.fit(X_train_poly4, y_train)

    train_r2 = r2_score(y_train, model.predict(X_train_poly4))
    test_r2 = r2_score(y_test, model.predict(X_test_poly4))

    print("Alpha", alpha, "| Train R2:", round(train_r2, 3), "| Test R2:", round(test_r2, 3))
```

***Find the Best Alpha (Ridge)***

```python
# RidgeCV tries each alpha using only the training data and picks the best one
ridge_cv = RidgeCV(alphas=[0.01, 0.1, 1, 10, 100, 1000])
ridge_cv.fit(X_train_poly4, y_train)

best_ridge_alpha = ridge_cv.alpha_
print("Best alpha for Ridge:", best_ridge_alpha)
```

***Train the Ridge Model***

```python
# Train Ridge with the best alpha
ridge_model = Ridge(alpha=best_ridge_alpha)
ridge_model.fit(X_train_poly4, y_train)

# Predict on train and test data
ridge_train_pred = ridge_model.predict(X_train_poly4)
ridge_pred = ridge_model.predict(X_test_poly4)

print("Train R2 :", round(r2_score(y_train, ridge_train_pred), 3))
print("Test R2  :", round(r2_score(y_test, ridge_pred), 3))
print("Test MAE :", round(mean_absolute_error(y_test, ridge_pred), 2))
print("Test RMSE:", round(np.sqrt(mean_squared_error(y_test, ridge_pred)), 2))
```

***Try Different Alpha Values (Lasso)***

```python
# Lasso can make some coefficients exactly 0
# max_iter=100000 gives the model enough steps to learn
for alpha in [1, 10, 50, 100, 300]:
    model = Lasso(alpha=alpha, max_iter=100000)
    model.fit(X_train_poly4, y_train)

    train_r2 = r2_score(y_train, model.predict(X_train_poly4))
    test_r2 = r2_score(y_test, model.predict(X_test_poly4))
    zero_count = (model.coef_ == 0).sum()

    print("Alpha", alpha, "| Train R2:", round(train_r2, 3),
          "| Test R2:", round(test_r2, 3), "| Zero coefficients:", zero_count)
```

***Find the Best Alpha (Lasso)***

```python
# LassoCV tries each alpha using only the training data and picks the best one
lasso_cv = LassoCV(alphas=[1, 10, 50, 100, 300], cv=5, max_iter=100000)
lasso_cv.fit(X_train_poly4, y_train)

best_lasso_alpha = lasso_cv.alpha_
print("Best alpha for Lasso:", best_lasso_alpha)
```

***Train the Lasso Model***

```python
# Train Lasso with the best alpha
lasso_model = Lasso(alpha=best_lasso_alpha, max_iter=100000)
lasso_model.fit(X_train_poly4, y_train)

# Predict on train and test data
lasso_train_pred = lasso_model.predict(X_train_poly4)
lasso_pred = lasso_model.predict(X_test_poly4)

print("Train R2 :", round(r2_score(y_train, lasso_train_pred), 3))
print("Test R2  :", round(r2_score(y_test, lasso_pred), 3))
print("Test MAE :", round(mean_absolute_error(y_test, lasso_pred), 2))
print("Test RMSE:", round(np.sqrt(mean_squared_error(y_test, lasso_pred)), 2))
```

***Prepare Data for Clustering***

```python
# Use all features, but NOT charges (clustering does not use the answer)
X_cluster = df_encoded.drop("charges", axis=1)

# Scale age, bmi and children (clustering uses distance, so scaling is needed)
scaler_cluster = StandardScaler()
X_cluster[num_cols] = scaler_cluster.fit_transform(X_cluster[num_cols])

print("Shape:", X_cluster.shape)
X_cluster.head()
```

***K-Means with 3 Clusters (First Try)***

```python
# Create K-Means model for 3 groups
kmeans3 = KMeans(n_clusters=3, random_state=42, n_init=10)

# Train the model and get the group number of each patient
labels3 = kmeans3.fit_predict(X_cluster)

# Count how many patients are in each group
print(pd.Series(labels3).value_counts().sort_index())
```

***Elbow Method***

```python

inertia_list = []

for k in range(1, 11):
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    model.fit(X_cluster)
    inertia_list.append(model.inertia_)

# Draw the graph
plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), inertia_list, marker="o")
plt.title("Elbow Method")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Inertia")
plt.show()
```

***Silhouette Score***

```python
# Silhouette score: higher is better (clusters are well separated)
silhouette_list = []

for k in range(2, 11):
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = model.fit_predict(X_cluster)
    silhouette_list.append(silhouette_score(X_cluster, labels))

# Draw the graph
plt.figure(figsize=(8, 5))
plt.plot(range(2, 11), silhouette_list, marker="o", color="green")
plt.title("Silhouette Score")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Silhouette Score")
plt.show()
```

***Choose Best k***

```python
# k with the highest silhouette score
best_k = silhouette_list.index(max(silhouette_list)) + 2

print("Best k by silhouette score:", best_k)
```

***Final K-Means with Best k***

```python
# Train K-Means again using the best k
kmeans_final = KMeans(n_clusters=best_k, random_state=42, n_init=10)
kmeans_labels = kmeans_final.fit_predict(X_cluster)

# Make a copy of the ORIGINAL data (with text values) and add the cluster number
df_clustered = df.copy()
df_clustered["KMeans_Cluster"] = kmeans_labels

# Count patients in each cluster
print(df_clustered["KMeans_Cluster"].value_counts().sort_index())
```
