# **Outlier**

```python
import pandas as pd 
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
#Generate a data set with potintail outlier
np.random.seed(0)
data = np.random.normal(100,20,200)
data = np.append(data,[300,5]) # Adding Potintial outlier
 # Calculate IQR 
 # Formula pf IQR is Q3-Q1 = IQR
Q1 = np.percentile(data,25)
Q3 = np.percentile(data,75)
IQR = Q3-Q1
 # Difining the outlier criteria
Minimum_outlier = Q1 - 1.5 * IQR
Maximum_outlier = Q3 + 1.5 * IQR
# identify the Outlier 
outlier = (data < Minimum_outlier) | (data > Maximum_outlier)
 # Removing outlier 
Clean_data = data[~outlier]

# ploting the original data with outlier 
plt.figure(figsize=(16,5))
plt.subplot(1,2,2)
sns.boxenplot(data=data)
plt.title('Orignal Data With Outlier')

# PLoting the data without outlier 
plt.subplot(1,2,1)
sns.boxplot(data=Clean_data)
plt.title('Data After Removing Outlier')
plt.show()
```
